```python
# STUDENT PERFORMANCE PREDICTION USING DEEP LEARNING
# Complete merged Python program
#
# Install required packages in Google Colab / terminal:
# !pip install tensorflow pandas numpy scikit-learn matplotlib
#
# Dataset: UCI Student Performance (Mathematics) dataset
# Target: G3 (final grade, 0-20)
# G1 and G2 are excluded from the input features so the model predicts
# performance without using the first- and second-period grades.

import os
import zipfile
import urllib.request
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score


# 1. REPRODUCIBILITY
SEED = 42
np.random.seed(SEED)
tf.random.set_seed(SEED)


# 2. DOWNLOAD AND LOAD DATASET
DATA_URL = (
    "https://archive.ics.uci.edu/ml/machine-learning-databases/00320/"
    "student.zip"
)
ZIP_FILE = "student.zip"
DATA_FILE = "student-mat.csv"

if not os.path.exists(DATA_FILE):
    print("Downloading UCI Student Performance dataset...")
    urllib.request.urlretrieve(DATA_URL, ZIP_FILE)
    with zipfile.ZipFile(ZIP_FILE, "r") as zip_ref:
        zip_ref.extract(DATA_FILE)

data = pd.read_csv(DATA_FILE, sep=";")

print("\nDataset loaded successfully.")
print("Dataset shape:", data.shape)
print(data.head())


# 3. SEPARATE INPUT FEATURES AND TARGET
# G3 is the final mathematics grade.
# G1 and G2 are removed to support an early prediction scenario.
target_column = "G3"
X = data.drop(columns=["G1", "G2", target_column])
y = data[target_column].astype("float32")

# Identify numerical and categorical columns.
numerical_columns = X.select_dtypes(include=["int64", "float64"]).columns.tolist()
categorical_columns = X.select_dtypes(include=["object", "category", "bool"]).columns.tolist()

print("\nNumber of numerical features:", len(numerical_columns))
print("Number of categorical features:", len(categorical_columns))


# 4. SPLIT DATA BEFORE FITTING PREPROCESSING
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=SEED
)

# OneHotEncoder compatibility with older and newer scikit-learn versions.
try:
    encoder = OneHotEncoder(handle_unknown="ignore", sparse_output=False)
except TypeError:
    encoder = OneHotEncoder(handle_unknown="ignore", sparse=False)

preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numerical_columns),
        ("cat", encoder, categorical_columns),
    ],
    sparse_threshold=0.0
)

X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

X_train_processed = np.asarray(X_train_processed, dtype=np.float32)
X_test_processed = np.asarray(X_test_processed, dtype=np.float32)

print("\nTraining data shape:", X_train_processed.shape)
print("Testing data shape:", X_test_processed.shape)


# 5. BUILD ARTIFICIAL NEURAL NETWORK
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(X_train_processed.shape[1],)),
    tf.keras.layers.Dense(64, activation="relu"),
    tf.keras.layers.Dropout(0.20),
    tf.keras.layers.Dense(32, activation="relu"),
    tf.keras.layers.Dense(16, activation="relu"),
    tf.keras.layers.Dense(1, activation="linear")
])

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss="mean_squared_error",
    metrics=["mean_absolute_error"]
)

print("\nMODEL SUMMARY")
model.summary()


# 6. TRAIN MODEL
early_stop = tf.keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=20,
    restore_best_weights=True
)

history = model.fit(
    X_train_processed,
    y_train,
    validation_split=0.20,
    epochs=200,
    batch_size=16,
    callbacks=[early_stop],
    verbose=1
)


# 7. EVALUATE MODEL
predictions = model.predict(X_test_processed, verbose=0).flatten()

mae = mean_absolute_error(y_test, predictions)
mse = mean_squared_error(y_test, predictions)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, predictions)

print("\nMODEL EVALUATION RESULTS")
print(f"Mean Absolute Error (MAE): {mae:.3f}")
print(f"Mean Squared Error (MSE): {mse:.3f}")
print(f"Root Mean Squared Error (RMSE): {rmse:.3f}")
print(f"R-squared (R2): {r2:.3f}")


# 8. DISPLAY ACTUAL VS PREDICTED VALUES
results = pd.DataFrame({
    "Actual_G3": y_test.to_numpy(),
    "Predicted_G3": predictions
})
print("\nACTUAL VS PREDICTED GRADES")
print(results.head(15).to_string(index=False))

results.to_csv("student_performance_predictions.csv", index=False)
print("\nPredictions saved to student_performance_predictions.csv")


# 9. PLOT TRAINING AND VALIDATION LOSS
plt.figure(figsize=(8, 5))
plt.plot(history.history["loss"], label="Training Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")
plt.title("Training and Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Mean Squared Error")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.savefig("training_validation_loss.png", dpi=300)
plt.show()


# 10. PLOT ACTUAL VS PREDICTED GRADES
plt.figure(figsize=(7, 6))
plt.scatter(y_test, predictions, alpha=0.7)
minimum = min(float(y_test.min()), float(predictions.min()))
maximum = max(float(y_test.max()), float(predictions.max()))
plt.plot([minimum, maximum], [minimum, maximum], linestyle="--")
plt.title("Actual vs Predicted Student Grades")
plt.xlabel("Actual Final Grade (G3)")
plt.ylabel("Predicted Final Grade (G3)")
plt.grid(True)
plt.tight_layout()
plt.savefig("actual_vs_predicted.png", dpi=300)
plt.show()


# 11. SAVE TRAINED MODEL
model.save("student_performance_model.keras")
print("\nTrained model saved as student_performance_model.keras")


# 12. PREDICT FOR A STUDENT FROM THE TEST SET
sample_features = X_test.iloc[[0]]
sample_actual = float(y_test.iloc[0])
sample_processed = preprocessor.transform(sample_features)
sample_processed = np.asarray(sample_processed, dtype=np.float32)
sample_prediction = float(model.predict(sample_processed, verbose=0)[0][0])

print("\nSAMPLE STUDENT PREDICTION")
print("Actual final grade:", sample_actual)
print("Predicted final grade:", round(sample_prediction, 2))

print("\nPROGRAM COMPLETED SUCCESSFULLY.")
```
